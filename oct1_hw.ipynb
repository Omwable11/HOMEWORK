```python
# 1. reverse a string

s = "Instagram"
rev = ""
# method 1
print(s[::-1])

# method 2
for ch in s:
    rev = ch + rev

print(rev)
```

```python
# 2. check palindrome

s = "madam"
rev = ""

# method 1
if s == s[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")

# method 2
for ch in s:
    rev = ch + rev

if s == rev:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
# 3. count vowels

s = "Instagram"
vowels = "AEIOUaeiou"
count = 0

for ch in s:
    if ch in vowels:
        count += 1

print(f"Your string contains {count} vowels")
```

```python
# 4. count consonants

s = "Instagram"
vowels = "AEIOUaeiou"
count = 0

for ch in s:
    if ch not in vowels:
        count += 1

print(f"Your string contains {count} consonants")
```

```python
# 6. find duplicates

s = "Programming"
seen = set()
duplicates = set()

for ch in s:
    if ch in seen:
        duplicates.add(ch)
    else:
        seen.add(ch)
print("Duplicates: ",duplicates)

```

```python
# 7. find first non-repeating character 

s = "aabbcdd"

# method 1
for ch in s:
    if s.count(ch) == 1:
        print(ch)
        break


# method 2
freq = {}
for ch in s:
    freq[ch] = freq.get(ch,0)+1

for ch in s:
    if freq[ch] == 1:
        print(ch)
        break
```

```python
# 8. remove duplicates

s = "Programming"
result = ""

for ch in s:
    if ch not in result:
        result += ch

print(result)
```

```python
# 9. check the strings are anagram or not 

a = 'lisTen'
b = 'Silent'

if sorted(a.lower()) == sorted(b.lower()):
    print("Anagram")
else:
    print("Not Anagram")

# method 2

if len(a) != len(b):
    print("Not Anagram")
else:
    freq = {}

    for char in a.lower():
        freq[char] = freq.get(char,0)+1

    for char in b.lower():
        freq[char] = freq.get(char,0)-1

    if all(value == 0 for value in freq.values()):
        print("Anagram")
    else:
        print("Not Anagram")
```

```python
# 10. frequency of each character

a = "banana"
freq = {}

for ch in a:
    if ch in freq:
        freq[ch] += 1
    else:
        freq[ch] = 1

print(freq)
```
