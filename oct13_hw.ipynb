```python
# print unique element list

l = [5,4,5,4,6,7,8,4,3,5,3,4]
unique_list = []

for ele in l:
    if ele not in unique_list:
        unique_list.append(ele)

print(unique_list)
```

```python
# count the frequency of the elements 

l = [5,4,5,4,6,7,8,4,3,5,3,4]

count_freq = {}

for ele in l:
    if ele in count_freq:
        count_freq[ele] += 1
    else:
        count_freq[ele] = 1

print(count_freq)
```

```python
# count frequency of character in the string

s = "I love python programming"

count_freq = {}

for char in s:
    if char in count_freq:
        count_freq[char] += 1
    else:
        count_freq[char] = 1

print(count_freq)
```

```python
db = {"name" : ["jay","raj","ria"],
      "marks" : [[78,67,"AB"],[98,"AB",81],["AB","AB",99]],
      "sub" : ["maths","python","english"]}

for i in range(len(db["name"])): #select student
    for j in range(len(db["marks"])): #select subject
        if db['marks'][i][j] == "AB": #absence
            print(f"Student {db["name"][i]} is absent in the subject {db["sub"][j]}")
```
