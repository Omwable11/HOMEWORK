```python
# task to display subject wise topper and congratulate them

db = {1:[88,82,90], 2:[78,80,85], 3:[92,88,95], 4:[70,75,80], 5:[85,90,88]}

phy_max = 0
chem_max = 0
math_max = 0

phy_roll = 0
chem_roll = 0
math_roll = 0

for k,v in db.items():
    if v[0] > phy_max:
        phy_max = v[0]
        phy_roll = k
    if v[1] > chem_max:
        chem_max = v[1]
        chem_roll = k
    if v[2] > math_max:
        math_max = v[2]
        math_roll = k

print(f"Congratulations ! Physics Topper is roll no. {phy_roll} and marks is {phy_max}.")
print(f"Congratulations ! Chemistry Topper is roll no. {chem_roll} and marks is {chem_max}.")
print(f"Congratulations ! Maths Topper is roll no. {math_roll} and marks is {math_max}.")
```
