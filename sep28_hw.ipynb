# Arithmetic Operators

```python
# + : Add 

a = 10
b = 12
print(a + b)

c = 10.2
d = 7.5
print(c + d)

x = "Python"
y = "Programming"
print(x +" "+ y)

m = 3 + 2j
n = 1 + 4j
print(m + n)

print(True + True)

print(False + False)

print(True + False)
```

```python
# - : subtract

print(10-4)

print(10.7 - 7.8)

print((5 + 3j) - (2 + 1j))

print(True - True)

print(True - False)
```

```python
# * : multiplication

print(10*4)

print(10.2*2.8)

print("python" * 5)

print((2 + 3j) * 2)

print(True * 5)

print(False * 5)
```

```python
# / : division

print(10/2)

print(2.7 / 9)

print((6+2j) / 2)

print(True / True)

print(False / True)
```

```python
# // : floor division

print(10 // 2)

print(10.5 // 3)

print(-10 // 3)
```

```python
# % : modulus

print(10 % 3)

print(10.5 % 3)
```

# Comparison Operator

```python
# == : equal to

print(10 == 10)

print(10.5 == 10.5)

print("Python" == "Python")

print("Python" == "Java")

print(True == False)

print((2+3j) == (2+3j))
```

```python
# != : not equal

print(10 != 5)

print(10.7 != 9.8)

print("Python" != "Python")

print(True != False)
```

```python
# > : greater than

print(10 > 2)

print(3.2 > 5.7)

```

```python
# < : less than

print(7 < 8)

print(5.2 < 8.8)
```

```python
#  >= : greater than or equal to

print(10 >= 10)

print(10 >= 4)

print(5 >= 9)

print(7.3 >= 9.2)

print(10.8 >= 9.8)
```

```python
# <= : less than or equal to

print(10 <= 12)

print(10 <= 8)

print(7.3 <= 9.5)
```
